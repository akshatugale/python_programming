```python
# Introduction to Python programming
''' 
Data Types of Python Programming

1. int : integer data type, used to represent whole numbers. (e.g., 1, 2, 3, -4, 0)
2. float : floating-point data type, used to represent decimal numbers. (e.g., 1.5, 2.7, -3.14)
3. str : string data type, used to represent text. anything written in quotes is considered as string. (e.g., "Hello, World!", 'Python','123')
4. bool : boolean data type, used to represent True or False values.
5. list : list data type, used to represent an ordered collection of items. (e.g., [1, 2, 3], ['apple', 'banana', 'cherry'])
6. tuple : tuple data type, used to represent an ordered collection of items that cannot be changed. tuple is immutable : we cant edit , remove, delete the values  (e.g., (1, 2, 3), ('apple', 'banana', 'cherry'))
7. set : set data type, used to represent an unordered collection of unique items. (e.g., {1, 2, 3}, {'apple', 'banana', 'cherry'})`
8. dict : dictionary data type, used to represent a collection of key-value pairs. (e.g., {'name': 'John', 'age': 30}, {'fruit': 'apple', 'color': 'red'})
9. NoneType : NoneType data type, used to represent the absence of a value or a null value. (e.g., None)
------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------
 
 #Data Structures of Python Programming

 1. list : mutable
 2. tuple : immutable
 3. dictonary : mutable
 4. sets : mutable

------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------
 '''

print('Hello World') 

name = 'akshat'
domain = 'DSDA'
experience = 15

#Formated String 
print(f'my name is {name} and domain {domain} , having exp is {experience} years')
```
