```python
#list in python programming 

lst = [1,2,3,4,5]
print(lst)
```

```python
#Accessing elements in a list 

#First element 
print(lst[0])

#Third element 
print(lst[2])

#Last element
print(lst[-1])
```

```python
#List methods are as follows:

#Modifying elements in a list 
lst[1] = 20
print(lst)
```

```python
#adding elemets to a lsit 
lst.append(6) 
print(lst)
```

```python
lst.insert(2 , 15)
print(lst)
```

```python
#Removing elemts from a list 
lst.remove(20)
print(lst)
```

```python
lst.pop(2)
print(lst)
```

```python
lst.count(5)
print(lst.count(5))
```

```python
lst.sort()
print(lst)
```

```python
lst.reverse()
print(lst)
```

```python
#List indexing and slicing 
my_list = [10,20,30,40,50]

#indexing 
print(my_list[0])
print(my_list[-1])
print(my_list[ ::1])
print(my_list[ ::2])
print(my_list[ ::-1])
```

```python
#Slicing 
print(my_list[1:4])
print(my_list[:3])
print(my_list[2:5])
```
